```python
%%writefile /kaggle/working/mvpruner-ablation.env
REPOSITORY=https://github.com/huyvanzzz/Log_Metrics.git
SOURCE_BRANCH=smolvlm-attention-nguyet
SOURCE_DIR=/kaggle/working/log-metrics-mvpruner
DATA_DIR=/kaggle/input/YOUR_DATASET/data
TEST_JSON="$DATA_DIR/multi_frame/multi_frame_test.json"
OUTPUT_DIR=/kaggle/working/mvpruner_ablation_outputs
BATCH_SIZE=8
```

```python
%%bash
set -euo pipefail
source /kaggle/working/mvpruner-ablation.env
if [ -d "$SOURCE_DIR/.git" ]; then
  test "$(git -C "$SOURCE_DIR" branch --show-current)" = "$SOURCE_BRANCH"
  test -z "$(git -C "$SOURCE_DIR" status --porcelain --untracked-files=no)"
  git -C "$SOURCE_DIR" pull --ff-only origin "$SOURCE_BRANCH"
else
  git clone --depth 1 --single-branch --branch "$SOURCE_BRANCH" "$REPOSITORY" "$SOURCE_DIR"
fi
python -m pip install -q -r "$SOURCE_DIR/smolvlm_attention/requirements.txt"
```

```python
%%bash
set -euo pipefail
source /kaggle/working/mvpruner-ablation.env
cd "$SOURCE_DIR"
python -m smolvlm_attention.mvpruner_ablation prepare \
  --test "$TEST_JSON" --image-root "$DATA_DIR" --output "$OUTPUT_DIR"
```

```python
%%bash
set -euo pipefail
source /kaggle/working/mvpruner-ablation.env
cd "$SOURCE_DIR"
python -u -m smolvlm_attention.mvpruner_ablation measure \
  --image-root "$DATA_DIR" --output "$OUTPUT_DIR" --batch-size "$BATCH_SIZE"
```

```python
%%bash
set -euo pipefail
source /kaggle/working/mvpruner-ablation.env
cd "$SOURCE_DIR"
python -m smolvlm_attention.mvpruner_ablation plot \
  --image-root "$DATA_DIR" --output "$OUTPUT_DIR"
```
